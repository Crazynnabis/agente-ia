```python
from agente_financiero.digestor_tecnico import ejecutar_ciclo_tecnico
from agente_financiero.agente_orb import ejecutar_analisis_orb
from agente_financiero.agente_petroleo import analizar_petroleo_completo
import asyncio, time

inicio = time.time()

# Corre los tres en paralelo
ciclo_tecnico, orb, petroleo = await asyncio.gather(
    ejecutar_ciclo_tecnico(),
    asyncio.to_thread(ejecutar_analisis_orb),
    analizar_petroleo_completo(),
)

duracion = round(time.time() - inicio, 1)
print(f"\nDuracion total: {duracion}s")
print(f"Señales tecnicas fuertes: {len(ciclo_tecnico['señales_fuertes'])}")
print(f"Señales ORB: {len([r for r in orb if r.get('señal') != 'ESPERAR'])}")
print(f"Petroleo WTI: ${petroleo['precios'].get('WTI', {}).get('precio', 'N/A')}")
```
