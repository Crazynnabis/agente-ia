```python
from firecrawl import FirecrawlApp
import ollama

app = FirecrawlApp(api_key="fc-9c31d8397e28409f8fbe8127a72f7df1")

# Extrae noticias
print("Extrayendo noticias...")
resultado = app.scrape("https://finance.yahoo.com", formats=["markdown"])
contenido = resultado.markdown[:3000]

# Analiza con Ollama
print("Analizando con IA local...")
respuesta = ollama.chat(
    model="llama3.2",
    messages=[
        {
            "role": "system",
            "content": "Eres un analista financiero. Responde siempre en español."
        },
        {
            "role": "user", 
            "content": f"Basándote en estas noticias financieras, dame 3 puntos clave del mercado hoy:\n\n{contenido}"
        }
    ]
)

print("\n=== ANÁLISIS DEL MERCADO ===")
print(respuesta["message"]["content"])
```
